```python
df = pd.read_csv('Dataset/raw_data.csv')
df1 = df.copy()
```

```python
df1
```

```python
df1.duplicated()

df1.drop_duplicates()
```

```python
df1[['name', 'country', 'gender']] = df1[['name', 'country', 'gender']].fillna('Unknown')
```

```python
df1
```

```python
# df1['name'] = df1['name'].str.split(' ')
```

```python
df1.ffill()

df1.bfill()
```

```python
age_mean = df1['age'].mean()
df1['age'] = df1['age'].fillna(age_mean).astype("int64")
```

```python
df1
```

```python
df1['country'] = df1['country'].replace('USA', 'United States')
```
