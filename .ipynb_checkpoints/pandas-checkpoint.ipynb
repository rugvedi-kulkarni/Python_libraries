# Query method

```python
# Query rule to filter data - returns a copy

# 1. Expressions as a string
# 2. Column based reference
# 3. Backticks for col_name with space or special chars
# 4. Operators (&, |, ~,>= <= <, ==, etc)
# 5. @to reference python vars
# 6. chaining
```

```python
df.query('aqi > 100')
df.query('aqi > 100 & temperature > 30')[['city', 'aqi']]
```

```python
aqi_val = 100
df.query('aqi > @aqi_val & temperature  > 30')[['city', 'aqi']]
```

# Transform Data (Feature Engineering)

```python
# Transforming data means creating some new columns, creating new features that improves the performance of our model

# 1. apply() - The logic to tranform our data we write it into fnx, and this fnx is passed to the apply fnx

df2 = df.copy()

df2['tax'] = df2['income'].apply(lambda x: '20%' if x >= 60000 else '10%')

df2
```

```python
# map function - when we map new val to existing val

df2 = df.copy()

gender_map = {
    "Male": 'M',
    'Female': 'F',
    "Unknown": 'U'
}

df2['gender'] = df2['gender'].map(gender_map)
```

```python
df2
```

```python
# assign function - create new columns by assigning new values to the existing values
df2.assign(new_income = df2['income'] * 1.1)
```

```python
# replace fnx - replace old to new val
df2['country'] = df2['country'].replace('USA', 'United States')
df2
```

```python
# Some other data tranformation methods

# Rename
df2.columns = ['Id', 'Name', 'Age', 'Country', 'Gender', 'Income']  # Rename all column names

df2.rename(columns = {
    'Income':'Salary'
}) # Rename specific column names

df2.rename(index = {2: 'Second'}) # rename label
```

```python
# sort_values() - column based
# suppose we sort values in terms of income

df2.sort_values('Income')

df2.sort_values('Income', ascending = False) # sort values in descending order

# values can also be sort on the basis of combination of two or more than 2 columns 
df2 = df2.fillna(50)
sorted_df = df2.sort_values(['Income', 'Age'])

sorted_df.sort_index() # sort in terms of index

sorted_df.reset_index(drop = True) # reset index and drop prev
```

```python
# Rank method - from lowest to highest values
# By default if there is a tie, if two values have the same rank then, there avg is given as a rank to both of them.
# here 1, is a lowest value
# min, max and dense - ranking values if two value clashes

sorted_df['Ranking'] = sorted_df['Income'].rank(ascending = False, method = 'dense')
```

```python
# re-order columns
sorted_df[['Id', 'Name', 'Country', 'Age','Gender', 'Income']]
```

```python
# Grouping and aggregation of Data
# groupby() - make grps of data

df.groupby('country')['income'].mean() # mean of countries accros income
df.groupby('country')['income'].min() # min income in a country
df.groupby('country')['income'].max() # max income in a country

# after grouping we apply aggregation fnx

df.groupby('gender')['income'].mean()
df.groupby('gender')['income'].min()
df.groupby('gender')['income'].max()
```

```python
# if we want to calculate mean(), max() and min()
df.groupby('country')['income'].agg(('mean', 'max', 'min')) 
df.groupby('country')['income'].aggregate(avg_salary = 'mean', max_salary = 'max', min_salary = 'min')
```

```python
# we can apply aggregate fnx to indivisual columns 
df.groupby('country').agg(
    {
       'income': 'mean',
        'age': 'mean'
    }
)

df.groupby('country').agg(
    {
       'income': 'max',
        'age': 'mean'
    }
)

# we can also rename values
df.groupby('country').agg(
    
      max_salary = ('income', 'max'),
      avg_age = ('age', 'mean')
    
)
```

```python
# melt and pivot
# when we transform our data from wide to long format it is called as melting.
# data visualization, transform our data
# we pass 4 parameters - id_vars => how many columns we want to KEEP
# 2. value_vars => values we want to un-pivot, values can be converted into single columns 
```

```python
df = pd.DataFrame({
    'country': ['USA', 'USA', 'India', 'India'],
    'year': [2020, 2021, 2020, 2021],
    'sales': [100, 120, 90, 110],
    'profit': [20, 25, 18, 22],
    # 'tax': [10, 20, 30, 40]
})

melted_df = df.melt(
    id_vars = ['country', 'year'],
    value_vars = ['sales', 'profit'],
    var_name = 'matrices',
    value_name = 'value'    
)

original = melted_df.pivot(
    index = ['country', 'year'],
    columns = 'matrices',
    values = 'value'
)
```

```python
# pivot - long to wide format (index, columns, values)
df
melted_df
original
```

```python
# Merging snd joining data

df_customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['Adam', 'Bob', 'Charlie', 'Dave']
})

df_orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104],
    'customer_id': [2, 1, 4, 5],
    'amount': [250, 120, 300, 100]
})
```

```python
df_customers
```

```python
df_orders
```

```python
pd.merge(df_customers, df_orders, on='customer_id') # inner join
pd.merge(df_customers, df_orders, on='customer_id', how = 'left') # left join
pd.merge(df_customers, df_orders, on='customer_id', how = 'right') # right join
pd.merge(df_customers, df_orders, on='customer_id', how = 'outer') # outer join
```

```python
# Data Concatenation
# Stacking DataFrames on top of each other (rows) or side by side (columns)

df1 = pd.DataFrame({
    'id': [1, 2, 3],
    'name': ['Adam', 'Eve', 'Bob']
})

df2 = pd.DataFrame({
    'id': [4, 5, 6],
    'name': ['Charlie', 'Fester', 'Gilian']
})

# row
pd.concat([df1, df2])
pd.concat([df2, df1], ignore_index = True)

# column
pd.concat([df1, df2], axis = 1)
```
