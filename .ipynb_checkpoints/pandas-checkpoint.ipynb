```python
s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))

# index - access
print(s[0])
print(s[1])
```

```python
s = pd.Series([23, 24, 25, 26], index = ["Adam", "Eve", "Bob", "Charvie"])
print(s)
print(s["Adam"])
print(s["Bob"])

print(s.index)
```

```python
# properties of series 
# 1. Homogenous 
# 2. Vectorization

s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
print(s1 + s2)

# 3. Memory optimization
# 4. Handle missing values with NaN

# 4. Mutable values, but size is immutable

s1[0] = 100
print(s1)

changed_s1 = s1.drop(0)
print(changed_s1) # give new series dont change the original series
```
