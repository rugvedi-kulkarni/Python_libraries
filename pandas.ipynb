# Query method

```python
# Query rule to filter data - returns a copy

# 1. Expressions as a string
# 2. Column based reference
# 3. Backticks for col_name with space or special chars
# 4. Operators (&, |, ~,>= <= <, ==, etc)
# 5. @to reference python vars
# 6. chaining
```

```python
df.query('aqi > 100')
df.query('aqi > 100 & temperature > 30')[['city', 'aqi']]
```

```python
aqi_val = 100
df.query('aqi > @aqi_val & temperature  > 30')[['city', 'aqi']]
```

# Transform Data (Feature Engineering)

```python
# Transforming data means creating some new columns, creating new features that improves the performance of our model

# 1. apply() - The logic to tranform our data we write it into fnx, and this fnx is passed to the apply fnx

df2 = df.copy()

df2['tax'] = df2['income'].apply(lambda x: '20%' if x >= 60000 else '10%')

df2
```

```python
# map function - when we map new val to existing val

df2 = df.copy()

gender_map = {
    "Male": 'M',
    'Female': 'F',
    "Unknown": 'U'
}

df2['gender'] = df2['gender'].map(gender_map)
```

```python
df2
```

```python
# assign function - create new columns by assigning new values to the existing values
df2.assign(new_income = df2['income'] * 1.1)
```

```python
# replace fnx - replace old to new val
df2['country'] = df2['country'].replace('USA', 'United States')
df2
```

```python
# Some other data tranformation methods

# Rename
df2.columns = ['Id', 'Name', 'Age', 'Country', 'Gender', 'Income']  # Rename all column names
df2.rename(columns = {'Income':'Salary'}) # Rename specific column names
df2.rename(index = {2: 'Second'}) # rename label
```

```python
# sort_values() - column based
# suppose we sort values in terms of income

df2.sort_values('Income')

df2.sort_values('Income', ascending = False) # sort values in descending order

# values can also be sort on the basis of combination of two or more than 2 columns 
df2 = df2.fillna(50)
sorted_df = df2.sort_values(['Income', 'Age'])

sorted_df.sort_index() # sort in terms of index

sorted_df.reset_index(drop = True) # reset index and drop prev
```
