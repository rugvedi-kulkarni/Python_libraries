```python
# Properties of series 
# 1. Homogenous 
# 2. Vectorization

s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
print(s1 + s2)

# 3. Memory optimization
# 4. Handle missing values with NaN

# 4. Mutable values, but size is immutable

s1[0] = 100
print(s1)

changed_s1 = s1.drop(0)
print(changed_s1) # give new series don't change the original series
```

```python
# AQI Data set - kaggle

import pandas as pd
import numpy as np

df = pd.read_csv('globalAirQuality.csv')

df.describe()
```

```python
# selecting data

# selecting columns
df[['city', 'aqi']] # access values of particular columns
```

```python
# select rows (loc and iloc) - indexer

print(df.loc[0])
print()
print(df.loc[0:2]) # start idx: end idx (inclusive)
print()
print(df.iloc[2])
print()
print(df.iloc[0:2]) # ending index exclusive
```

```python
# select rows and columns
print(df.loc[0, 'aqi']) # why np is printed, because we use numpy in pandas

print(df.loc[0, ['city', 'aqi']])

print(df.loc[0:2, ['city', 'aqi']])

print(df.columns)

print(df.iloc[0:3, 2]) # iloc requires numeric index of column

print(df.loc[0:2, ['country', 'city', 'latitude', 'longitude']])

print(df.iloc[0:3, 1:5])
```

```python
# select single scalar value - at(label) and iat (index)

print(df.at[0, 'city'],'\n',
df.iat[0, 2])
```

```python
# all the functions returns a view not copy
# remember if we want to make changes in original data then make a copy of it and then make a change

cities = df['city'].copy()
cities[0] = 'Mumbai'
```
